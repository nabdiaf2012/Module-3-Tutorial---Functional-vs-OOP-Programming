```python
#1. Sort an Array of 0s, 1s, and 2s

class Solution:
    def sort012(self, arr):
        low = 0
        mid = 0
        high = len(arr) - 1

        while mid <= high:
            if arr[mid] == 0:
                arr[low], arr[mid] = arr[mid], arr[low]
                low += 1
                mid += 1

            elif arr[mid] == 1:
                mid += 1

            else:
                arr[mid], arr[high] = arr[high], arr[mid]
                high -= 1

        return arr


# Test the solution
arr = [0, 1, 2, 0, 1, 2]
solution = Solution()
print(solution.sort012(arr))
```

```python
#2. Binary Search

class Solution:
    def binarysearch(self, arr, k):
        low = 0
        high = len(arr) - 1

        while low <= high:
            mid = (low + high) // 2

            if arr[mid] == k:
                return mid

            elif arr[mid] < k:
                low = mid + 1

            else:
                high = mid - 1

        return -1


# Test the solution
arr = [1, 2, 3, 4, 5]
k = 4

solution = Solution()
print(solution.binarysearch(arr, k))
```
